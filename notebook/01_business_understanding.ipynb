#  Project Objective

The objective of this project is to build an intelligent fraud detection system capable of identifying fraudulent credit card transactions using machine learning techniques.

Since fraudulent transactions represent only a very small fraction of all transactions, the dataset is highly imbalanced. Therefore, the goal is not only to achieve high accuracy but also to maximize fraud detection while minimizing false alarms.

The final system will include experiment tracking using MLflow, model explainability using SHAP, and an interactive prediction interface using Streamlit.

## Business Problem

Banks process millions of transactions every day.

A fraudulent transaction causes direct financial loss.

A false alarm blocks a genuine customer's transaction and damages customer experience.

Therefore the model should maximize fraud detection while reducing unnecessary transaction blocks.
